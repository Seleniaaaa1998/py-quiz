```python
# write a program that adds the numbers from 1 to 10
sum = 0 # initialize a variable sum to 0 : it will accumulate the result
for i in range(1, 11): # starts a loop where i takes integer from 1 to 10
    sum += i # on each iteration, it adds the current i to sum
print("sum is:", sum) # prints message with result
```

```python
# write a program that distinguishes whether a number is even or odd
num = int(input("Enter a number: ")) # asks the user to input a number and int converts the input from strings to integer
if num % 2 == 0: # checks if the number is even
    print(num, "is an even number.")
else: # if the number is not even, it must be odd
    print(num, "is an odd number.")

```

```python
# write a list that calculates the average of the grades of an exam saved
n = int(input("How many grades do you want to enter? ")) # asks the user how many grades they want to enter. the answer is converted to an integer and the result is saved in n
sum = 0 # initializes a variable sum to 0. it is used to add all the grades entered (sum grades)
for i in range(n): # creates a loop that repeats n times, that is, for each grade the user wants to enter
    grade = int(input("Enter the grade: ")) #in each iteration, it asks the user to enter a grade, and converts it to an integer
    sum += grade # adds the newly entered grade to the total sum
average =sum / n # calculates the average of the grades by dividing the total sum by the number of grades entered
print("The average of the grades is:", average) # prints the calculated average
```

```python
# write a listing that tests whether a list is sorted in ascending order
list = [1, 2, 3, 4, 5]  # create a list of numbers
is_sorted = True  # start by assuming the list is sorted

for i in range(len(list) - 1):  
    # loop through the list up to the second-to-last element
    # we stop at len(list)-1 because we compare each element with the next one (i+1)
    if list[i] > list[i + 1]:  
        # if current element is greater than the next, list is not sorted
        is_sorted = False
        break  # exit the loop immediately since list is not sorted

if is_sorted:
    print("The list is sorted.")  # print if list is sorted in ascending order
else:
    print("The list is not sorted.")  # print if list is not sorted



#for i in range(len(list)):
#    print(list[i])
#list = [1, 2, 3, 4, 5]
#print(list[1])  # Stampa solo il numero 2
```
